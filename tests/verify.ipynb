```python
import numpy as np
from collections import Counter

def load_matrix(filename):
    with open(filename, 'rb') as f:
        # 读取矩阵大小
        size = np.fromfile(f, dtype=np.int64, count=1)[0]
        # 读取矩阵数据
        data = np.fromfile(f, dtype=np.int32)
    return data

def find_most_frequent(arr):
    # 使用Counter计算频率
    counter = Counter(arr)
    # 找到最频繁的元素
    most_common = counter.most_common(1)[0]
    return most_common[0], most_common[1]
```

```python
matrix = load_matrix('test_matrix.bin')
```

```python
print("Loading matrix from file...")

print("\nCalculating most frequent element...")
value, count = find_most_frequent(matrix)
print(f"Most frequent element: {value}")
print(f"Frequency: {count}")

# 计算一些基本统计信息
print("\nBasic statistics:")
print(f"Unique values count: {len(np.unique(matrix))}")
print(f"Min value: {np.min(matrix)}")
print(f"Max value: {np.max(matrix)}")
```
